### Prepara los datos

Ejecuta estas dos celdas. Dejan lista `postulaciones`, con `3200` filas, y la función `llamados`.

Recuerda: **el sesgo está programado a propósito**. Este archivo no dice nada de personas reales.

```python
import pandas as pd

DATOS = "https://raw.githubusercontent.com/PercyTechX/del-excel-al-ml/main/datos/"
```

```python
def llamados(tabla):
    """Por cargo y sexo: a cuántas personas se llamó, de cuántas, y qué parte."""
    return tabla.pivot_table(
        index="cargo",
        columns="sexo",
        values="entrevistado",
        aggfunc=["sum", "size", "mean"],
    ).round(2)


postulaciones = pd.read_csv(DATOS + "postulaciones.csv")

print(len(postulaciones))
```

### 1. Por sede

Calcula, por `sede` y `sexo`, la parte llamada a entrevista, entre quienes sacaron 75 o más. ¿En qué sede hay menos casos de mujeres?

```python
altos = postulaciones[postulaciones["puntaje_prueba"] >= 75]

altos.pivot_table(
    index="sede", columns="sexo", values="entrevistado", aggfunc=["size", "mean"]
).round(2)

# La sede con menos mujeres es Cusco: 19.
```

### 2. El total

¿A cuántas mujeres y a cuántos hombres se llamó en 2025? Deben ser `35` y `180`, de `271` y `512`.

```python
de_2025 = postulaciones[postulaciones["fecha"].str[:4] == "2025"]

de_2025.groupby("sexo")["entrevistado"].agg(["sum", "size"])
```

### 3. Un puntaje más bajo

Repite la tabla por cargo con `llamados()`, entre quienes sacaron de 50 a 74. Filtra con `.between(50, 74)`. ¿Se repite la dirección?

```python
medios = postulaciones[postulaciones["puntaje_prueba"].between(50, 74)]

llamados(medios)

# Sí, la dirección se repite.
```

### 4. Los meses sin empleo

Entre las personas de Asistente administrativo, calcula la parte llamada según tengan 12 meses sin empleo o más, o menos de 12.

> 💡 `tabla.groupby(tabla["meses_sin_empleo"] >= 12)["entrevistado"].mean()`. Debe dar `0.297` si tiene menos de 12, y `0.115` si tiene 12 o más.

```python
asistentes = postulaciones[postulaciones["cargo"] == "Asistente administrativo"]

asistentes.groupby(asistentes["meses_sin_empleo"] >= 12)["entrevistado"].mean().round(3)
```
