### 1. La factura

Distribuidora Andina vende **12 cajas de Aceite vegetal 1 L** a **S/ 105** cada una. Crea las variables `precio` y `cantidad`, y calcula `valor_venta`, `igv` (18 %) y `total`. Muestra el total redondeado a 2 decimales.

```python
precio = 105
cantidad = 12

valor_venta = precio * cantidad
igv = valor_venta * 0.18
total = valor_venta + igv

# 1486.8 y 1486.80 son el mismo número; el ejercicio 2 muestra los dos decimales
round(total, 2)
```

```python
print(f"Total de la factura: S/ {total:,.2f}")
```

### 4. ¿Cuánto tiempo lleva el cliente?

*Bodega Doña Karina* es cliente desde el **15 de setiembre de 2021**. Calcula cuántos **días** pasaron hasta el **31 de diciembre de 2025**, y muestra el **año** de su fecha de alta.

> 💡 Pista: empieza con `from datetime import date`.

```python
from datetime import date

fecha_alta = date(2021, 9, 15)
fecha_final = date(2025, 12, 31)

print((fecha_final - fecha_alta).days)
print(fecha_alta.year)
```

### 7. La ficha del producto

Crea un diccionario `producto` con estos datos: `id_producto` = `"P007"`, `nombre` = `"Atún en trozos 170 g"`, `precio` = `118` y `costo` = `101.14`.

1. Muestra el nombre.
2. Calcula el margen (precio menos costo), redondeado a 2 decimales.
3. Cambia el precio a `125` y calcula el margen nuevo.
4. Agrega el dato `marca` = `"Puerto Azul"` y muestra el diccionario completo.

```python
producto = {
    "id_producto": "P007",
    "nombre": "Atún en trozos 170 g",
    "precio": 118,
    "costo": 101.14,
}

print(producto["nombre"])
print(round(producto["precio"] - producto["costo"], 2))

producto["precio"] = 125
print(round(producto["precio"] - producto["costo"], 2))

producto["marca"] = "Puerto Azul"
print(producto)
```

### 8. Caza de errores

Estas son las celdas corregidas. Cada una indica qué error daba y a qué error de Excel se parece.

**a)** `TypeError`, como `#¡VALOR!`: se intentó sumar un texto y un número. Excel habría convertido este texto por su cuenta; Python no, así que hay que usar `float()`.

```python
venta = "1250"

float(venta) + 100
```

**b)** `NameError`, como `#¿NOMBRE?`: `totla` está mal escrito.

```python
total = 1250 + 980

total
```

**c)** `IndexError`, como `#¡REF!`: la lista tiene 3 valores y se cuenta desde 0, así que el último está en la posición `2`. También vale `ventas[-1]`.

```python
ventas = [1250.50, 980, 1430]

ventas[2]
```

**d)** `KeyError`, como `#N/D`: el dato se llama `"precio"`, en minúsculas, y `"Precio"` no existe. Ojo: `=BUSCARV()` no distingue mayúsculas; Python sí.

```python
producto = {"nombre": "Atún en trozos 170 g", "precio": 118}

producto["precio"]
```
