### Paso 1: variables, tus celdas con nombre

Una **variable** es un nombre que guarda un valor. Es como una celda con nombre en Excel.

Vamos a vender 3 fardos de *Arroz extra 5 kg*, que cuestan S/ 99.50 cada uno:

```python
precio = 99.5
cantidad = 3

precio * cantidad
```

Aquí el `=` no significa "es igual a", sino **"guarda esto"**: `precio = 99.5` guarda `99.5` con el nombre `precio`.

Los nombres de variable siguen tres reglas:

| Regla | ✅ Bien | ❌ Mal |
|---|---|---|
| Sin espacios (usa `_`) | `valor_venta` | `valor venta` |
| No empiezan con un número | `ventas_2025` | `2025_ventas` |
| Mayúsculas y minúsculas cuentan | `precio` y `Precio` son **dos variables distintas**, a diferencia de Excel | |

También podemos guardar el resultado. Si la última línea de una celda es un cálculo o un nombre, Colab muestra su valor. Para mostrar un valor en cualquier otra línea, usa `print()`.

```python
total = precio * cantidad
print(total)
```

### ⚠️ La gran diferencia con Excel

En Excel, si cambias el precio, el total se recalcula solo. **En Python, no.** Mira qué pasa si el precio sube a S/ 105:

```python
precio = 105.0

total
```

¡El total sigue igual! `total` guardó **el resultado** de la cuenta, no la fórmula. Para actualizarlo, hay que volver a calcularlo:

```python
total = precio * cantidad

total
```

### Paso 2: los tipos de datos

En Excel, una celda puede tener un número, un texto o una fecha. En Python es igual, y a cada tipo se le da un nombre:

| Dato | Ejemplo | Tipo en Python |
|---|---|---|
| Número entero | `3` | `int` |
| Número con decimales | `99.5` | `float` |
| Texto (siempre entre comillas) | `"Arroz extra 5 kg"` | `str` |
| Fecha | `date(2024, 1, 2)` | `date` |

Para saber qué tipo tiene un valor, usa `type()`:

```python
nombre = "Arroz extra 5 kg"

type(nombre)
```

#### Textos: unir palabras y números

En Excel unes textos con `&`: `=A1&": S/ "&B1`. En Python se usa un **texto con `f`**: pones una `f` antes de las comillas y cada variable entre llaves `{ }`.

```python
f"{nombre}: S/ {total}"
```

Las comillas de la salida solo indican que es un texto. Con `print()` desaparecen.

#### Cuidado con los números guardados como texto

Seguro te ha pasado en Excel: un número que está como texto y no suma bien. Con comillas, `"99.50"` es un **texto**. Mira qué hace Python si lo multiplicas por 2:

```python
precio_texto = "99.50"

precio_texto * 2
```

¡Repitió el texto dos veces, y sin avisar! Para convertirlo en número, usa `float()`, como `=VALOR()` (`VALUE`) en Excel:

```python
float(precio_texto) * 2
```

#### Redondear y dar formato

Te lo prometimos en el episodio 0.1. `round()` funciona como `=REDONDEAR()` (`ROUND`): le das el número y la cantidad de decimales, separados por coma. (Si tu Excel usa `;` entre argumentos, en Python siempre es `,`).

```python
promedio = (1250.50 + 980 + 1430) / 3

print(promedio)
print(round(promedio, 2))
```

> ⚠️ `round()` no siempre coincide con Excel en los casos que terminan en 5: `round(2.675, 2)` da `2.67`, y `=REDONDEAR(2.675,2)` da `2.68`. Tenlo en cuenta si tus números deben cuadrar exactamente con un reporte de Excel.

`round()` tampoco muestra los ceros finales: `1486.80` aparece como `1486.8`. Para **mostrar** un monto con sus dos decimales, como con *Formato de celdas* en Excel, agrega `:,.2f` dentro de las llaves. Significa: con separador de miles (`,`) y 2 decimales (`.2f`).

```python
total_vendido = 12580.5

print(f"Total vendido: S/ {total_vendido:,.2f}")
```

#### Fechas

En Excel usas `=FECHA(2022,5,5)` (`DATE`). En Python, la herramienta de fechas se trae con `import`, una línea que basta ejecutar una vez por sesión:

```python
from datetime import date

fecha_alta = date(2022, 5, 5)
fecha_venta = date(2024, 1, 2)

print(fecha_venta)
print((fecha_venta - fecha_alta).days)
print(fecha_venta.year)
```

Se escribe en orden **año, mes, día**. Igual que en Excel, puedes restar dos fechas para saber cuántos días hay entre ellas.

El punto significa "de este valor, dame...": la resta da una duración y `.days` saca sus días (va entre paréntesis para restar primero). `.year` y `.month` funcionan como `=AÑO()` (`YEAR`) y `=MES()` (`MONTH`).

### Paso 3: listas, tus columnas

Una **lista** guarda varios valores en orden, como una columna de Excel. Se escribe entre corchetes `[ ]`, con los valores separados por comas:

```python
ventas = [1250.50, 980, 1430, 1100, 875.50]

ventas
```

Muchas funciones de Excel tienen su equivalente directo:

| En Excel | En Python |
|---|---|
| `=SUMA()` (`SUM`) | `sum(ventas)` |
| `=CONTARA()` (`COUNTA`) | `len(ventas)` |
| `=MAX()` | `max(ventas)` |
| `=MIN()` | `min(ventas)` |
| `=PROMEDIO()` (`AVERAGE`) | `sum(ventas) / len(ventas)` |

```python
print(sum(ventas))
print(len(ventas))
print(max(ventas))
print(round(sum(ventas) / len(ventas), 2))
```

Para sacar un valor, usa su **posición** entre corchetes. Ojo: **en Python se empieza a contar desde 0**, no desde 1. Y `-1` significa "el último".

```python
print(ventas[0])  # el primero
print(ventas[-1])  # el último
```

Para agregar un valor al final, como escribir en la primera fila vacía de la columna, usa `.append()`:

> ⚠️ Cada vez que ejecutes esta celda se agrega otra vez. Si se duplicó, vuelve a ejecutar la celda que crea la lista.

```python
ventas.append(1640)

ventas
```

### Paso 4: diccionarios, tus fichas

Un **diccionario** guarda los datos de una sola cosa, como una fila de Excel. Cada dato tiene un nombre, igual que el encabezado de una columna. Se escribe entre llaves `{ }`, con cada dato como `"nombre": valor`, separados por comas. Ese nombre se llama **clave** (*key* en inglés).

```python
producto = {
    "id_producto": "P001",
    "nombre": "Arroz extra 5 kg",
    "precio": 99.5,
    "costo": 85.21,
}

producto["nombre"]
```

En la lista buscas por **posición**. En el diccionario buscas por **nombre**, así que no tienes que recordar en qué orden está cada dato.

Con esos datos también puedes calcular, cambiar un valor o agregar uno nuevo:

> 💡 A veces Python muestra algo como `19.790000000000006`. Es normal en cálculos con decimales; por eso redondeamos.

```python
producto["precio"] = 105.0  # cambia un dato que ya existe
producto["marca"] = "Valle Dorado"  # agrega un dato nuevo

margen = producto["precio"] - producto["costo"]

print(round(margen, 2))
print(producto)
```

### Paso 5: leer un error sin miedo

Un error no significa que rompiste algo. Es Python diciéndote qué no entendió, igual que Excel cuando muestra `#¡VALOR!`.

**El truco: empieza por la última línea.** Ahí está el tipo de error y su explicación. Las líneas de arriba, con una flecha `---->`, indican dónde ocurrió.

Las siguientes celdas fallan a propósito. Ejecútalas una por una.

**Dividir entre cero**, el `#¡DIV/0!` de Python:

```python
1250 / 0
```

```python
precio_texto + 5
```

```python
ventas[10]
```

| En Excel | En Python | Qué significa |
|---|---|---|
| `#¡DIV/0!` | `ZeroDivisionError` | Dividiste entre cero |
| `#¡VALOR!` | `TypeError` | Mezclaste tipos de datos, por ejemplo, texto y número |
| `#¿NOMBRE?` | `NameError` | Python no conoce ese nombre: revisa cómo lo escribiste y si ejecutaste la celda que lo crea |
| `#¡REF!` | `IndexError` | Esa posición no existe en la lista |
| `#N/D` | `KeyError` | Ese dato no existe en el diccionario |

## ✅ Resumen

| En Excel | En Python |
|---|---|
| Celda con nombre | Variable: `precio = 99.5` |
| Se recalcula solo | Hay que volver a calcular |
| Una columna | Lista: `[1250.50, 980, 1430]` |
| Una fila o ficha | Diccionario: `{"nombre": "Arroz", "precio": 99.5}` |
| `=SUMA()` (`SUM`), `=CONTARA()` (`COUNTA`) | `sum()`, `len()` |
| `=REDONDEAR()` (`ROUND`) | `round()` |
| `=VALOR()` (`VALUE`) | `float()` |
| `=FECHA()` (`DATE`) | `date()` |
| `&` para unir textos | `f"{nombre}: S/ {total}"` |
