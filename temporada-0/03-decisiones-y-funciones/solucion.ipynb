### 1. ¿Se aprueba el pedido?

*Bodega Doña Karina* tiene un límite de crédito de **S/ 8,000** y hace un pedido de **S/ 9,200**. Crea las variables `limite_credito` y `pedido`, y muestra `Excede el límite` si el pedido es **mayor** que el límite, o `Pedido aprobado` si no lo es.

Luego cambia el pedido a **S/ 8,000**, vuelve a ejecutar la celda y comprueba qué mensaje sale.

```python
limite_credito = 8000
pedido = 9200

if pedido > limite_credito:
    print("Excede el límite")
else:
    print("Pedido aprobado")
```

### 3. Envío gratis

El envío es gratis si la ciudad es **Lima** o si el total es de **S/ 2,000 o más**. Crea las variables `ciudad` = `"Arequipa"` y `total` = `2500`, y muestra `Envío gratis` o `Envío con costo`.

### 4. Precios con IGV

Ejecuta la celda de abajo para crear la lista `precios`. En la siguiente, crea la lista `precios_con_igv` con cada precio más el 18 %, redondeado a 2 decimales, y muéstrala.

```python
precios = [99.5, 42.0, 39.0, 105.0, 56.0]
```

```python
precios_con_igv = []

for precio in precios:
    precios_con_igv.append(round(precio * 1.18, 2))

precios_con_igv
```

### 5. Las ventas grandes

Ejecuta la celda de abajo para crear la lista `ventas`. En la siguiente, calcula cuánto suman las ventas de **S/ 1,000 o más** y **cuántas** son.

```python
ventas = [1250.50, 980, 1430, 5200, 875.50, 1640]
```

```python
suma_grandes = 0
cantidad_grandes = 0

for venta in ventas:
    if venta >= 1000:
        suma_grandes = suma_grandes + venta
        cantidad_grandes = cantidad_grandes + 1

print(suma_grandes)
print(cantidad_grandes)
```

### 6. Tu propia fórmula

Crea la función `calcular_margen`, que reciba el `precio` y el `costo`, y entregue el precio menos el costo, redondeado a 2 decimales.

> 💡 Pista: si una función recibe dos datos, se separan con coma: `def calcular_margen(precio, costo):`.

Pruébala con el *Arroz extra 5 kg* (precio `99.5`, costo `85.21`) y con el *Atún en trozos 170 g* (precio `118`, costo `101.14`).

```python
def calcular_margen(precio, costo):
    return round(precio - costo, 2)


print(calcular_margen(99.5, 85.21))
print(calcular_margen(118, 101.14))
```

### 7. Todo junto

Crea la función `clasificar_pedido`, que reciba un `total` y entregue el texto `"Grande"`, `"Mediano"` o `"Pequeño"`, con las reglas del ejercicio 2. Después, recorre la lista `ventas` del ejercicio 5 y muestra una línea por cada venta, con este formato:

`S/ 1,250.50: Mediano`

> 💡 Si te sale `None`, a tu función le falta un `return`. Si te sale `NameError`, vuelve a ejecutar la celda que crea `ventas`.

```python
def clasificar_pedido(total):
    if total >= 5000:
        return "Grande"
    elif total >= 1000:
        return "Mediano"
    else:
        return "Pequeño"


for venta in ventas:
    categoria = clasificar_pedido(venta)
    print(f"S/ {venta:,.2f}: {categoria}")
```
