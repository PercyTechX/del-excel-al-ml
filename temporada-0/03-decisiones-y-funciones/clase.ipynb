## 🐍 En Python

### Paso 1: decisiones con `if`

`=SI(condición, si es verdad, si no)` se escribe así en Python:

```python
total = 1250.50

if total >= 1000:
    print("Aplica descuento")
else:
    print("Sin descuento")
```

Fíjate en dos detalles que no existen en Excel:

- Los **dos puntos** `:` al final de la línea del `if` y del `else`.
- La **sangría**: los espacios al inicio de la línea. Así Python sabe qué líneas dependen de la condición. Colab la agrega al presionar Enter después de los dos puntos, y puede usar 2 o 4 espacios: ambos funcionan, pero usa el mismo ancho en todo el bloque. Para salir del bloque, borra esos espacios.

Si olvidas la sangría, verás un `IndentationError`. Si olvidas los dos puntos o usas `=` en vez de `==`, verás un `SyntaxError`. Lee la última línea: suele decirte qué falta.

Las comparaciones casi son las de Excel. Cambian dos:

| Pregunta | En Excel | En Python |
|---|---|---|
| ¿Es igual? | `=` | `==` (con **dos** signos) |
| ¿Es distinto? | `<>` | `!=` |
| Mayor, menor | `>`, `<` | `>`, `<` |
| Mayor o igual, menor o igual | `>=`, `<=` | `>=`, `<=` |

Recuerda: un solo `=` significa "guarda esto", y `==` pregunta "¿son iguales?".

### Más de dos caminos: `elif`

Lo que en Excel harías con varios `=SI()` anidados o con `=SI.CONJUNTO()` (`IFS`), en Python se hace con `elif`. Python revisa de arriba hacia abajo y se queda con **la primera condición que se cumpla**. Por eso el orden importa: empieza por la más exigente.

```python
if total >= 5000:
    print("Pedido grande")
elif total >= 1000:
    print("Pedido mediano")
else:
    print("Pedido pequeño")
```

### Dos condiciones a la vez: `and` y `or`

`=Y()` (`AND`) es `and`: se deben cumplir **las dos**. `=O()` (`OR`) es `or`: basta con **una**.

```python
ciudad = "Lima"

if ciudad == "Lima" and total >= 1000:
    print("Envío gratis")
else:
    print("Envío con costo")
```

```python
ciudad = "Cusco"

if ciudad == "Lima" or total >= 1000:
    print("Envío gratis")
else:
    print("Envío con costo")
```

> ⚠️ Python distingue mayúsculas al comparar textos: `"lima" == "Lima"` no se cumple. En Excel, sí.

### Paso 2: repetir con `for`

En Excel arrastras la fórmula hacia abajo. En Python, `for` recorre la lista y hace el cálculo una vez por cada valor:

```python
ventas = [1250.50, 980, 1430, 1100, 875.50]

for venta in ventas:
    print(round(venta * 1.18, 2))
```

Se lee así: **"por cada `venta` en `ventas`, haz esto"**. El nombre `venta` lo eliges tú, y en cada vuelta guarda un valor distinto. Aquí también van los dos puntos y la sangría.

Para guardar los resultados en una **columna nueva**, empieza con una lista vacía y agrega cada resultado con `.append()`:

```python
ventas_con_igv = []

for venta in ventas:
    ventas_con_igv.append(round(venta * 1.18, 2))

ventas_con_igv
```

### Sumar y contar con condiciones

`=SUMAR.SI()` (`SUMIF`) y `=CONTAR.SI()` (`COUNTIF`) se arman juntando `for` e `if`. Empezamos en cero y vamos sumando. Una variable que va sumando así se llama **acumulador**:

```python
suma_grandes = 0
cantidad_grandes = 0

for venta in ventas:
    if venta >= 1000:
        suma_grandes = suma_grandes + venta
        cantidad_grandes = cantidad_grandes + 1

print(suma_grandes)
print(cantidad_grandes)
```

`suma_grandes = suma_grandes + venta` significa: "calcula lo de la derecha y guárdalo con ese nombre". En Excel sería una referencia circular; en Python es normal.

Y mira la sangría: el `if` está dentro del `for`, y las sumas están dentro del `if`.

### Paso 3: tus propias fórmulas con `def`

Si repites un cálculo muchas veces, conviértelo en una **función**. Es como crear tu propia fórmula de Excel:

```python
def calcular_igv(valor_venta):
    return round(valor_venta * 0.18, 2)
```

- `def` crea la función, y `calcular_igv` es su nombre.
- `valor_venta` es el dato que recibe, como lo que escribes entre los paréntesis de `=REDONDEAR()` (`ROUND`).
- `return` entrega el resultado para que lo uses después. `print` solo lo muestra en pantalla. Si ves `None`, a tu función le falta un `return`.

Esa celda no muestra nada: solo **guarda** la fórmula. Para usarla, escribe su nombre con el dato entre paréntesis:

```python
calcular_igv(1500)
```

> 💡 Si cambias algo dentro de la función, vuelve a ejecutar su celda. Si no, Python sigue usando la versión anterior.

### Todo junto

La regla de la promoción, guardada en una función y aplicada a toda la lista:

```python
def calcular_descuento(monto):
    if monto >= 1000:
        return round(monto * 0.05, 2)
    else:
        return 0


for venta in ventas:
    descuento = calcular_descuento(venta)
    print(f"Venta: S/ {venta:,.2f} | Descuento: S/ {descuento:,.2f}")
```

## ✅ Resumen

| En Excel | En Python |
|---|---|
| `=SI()` (`IF`) | `if` / `else` |
| `=SI.CONJUNTO()` (`IFS`) | `if` / `elif` / `else` |
| `=Y()` (`AND`), `=O()` (`OR`) | `and`, `or` |
| `=` y `<>` para comparar | `==` y `!=` |
| Arrastrar la fórmula hacia abajo | `for` |
| `=SUMAR.SI()` (`SUMIF`), `=CONTAR.SI()` (`COUNTIF`) | `for` con `if` y un acumulador |
| Una fórmula que reutilizas | Una función con `def` |
