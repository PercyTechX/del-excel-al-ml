## 🐍 En Python

### Paso 1: abrir un archivo

**pandas** es la herramienta de Python para trabajar con tablas. Colab ya la trae instalada; solo hay que llamarla con `import`. Le decimos `pd` para escribir menos.

Abajo también guardamos la dirección de internet donde están los datos del curso:

````python
import pandas as pd

DATOS = "https://raw.githubusercontent.com/PercyTechX/del-excel-al-ml/main/datos/"
````

Un **CSV** es una tabla guardada como texto, con los datos separados por comas. `pd.read_csv()` lo abre, y lo guardamos en una variable. Con `+` unimos dos textos: la dirección y el nombre del archivo. A esta tabla, pandas la llama **DataFrame**: es como una hoja de Excel.

`.head()` muestra las primeras 5 filas y `.shape` dice cuántas filas y columnas tiene. `.shape` es un dato y `.head()` una acción: por eso solo esta lleva paréntesis.

````python
ventas = pd.read_csv(DATOS + "ventas.csv")

print(ventas.shape)
ventas.head()
````

> 💡 Los números en negrita de la izquierda son el **índice**: identifican cada fila, como los números de fila de Excel. Empiezan en 0, igual que las listas.

Para un libro de Excel, usa `pd.read_excel()` e indica la hoja con `sheet_name`. Sin esa opción, abre la primera hoja. Dentro de los paréntesis, `=` indica una opción, no guarda nada.

````python
productos = pd.read_excel(DATOS + "reporte_ventas_2025.xlsx", sheet_name="Productos")

productos.head()
````

### Paso 2: elegir columnas

Una columna se pide como un dato de un diccionario, con su nombre entre corchetes:

````python
ventas["ciudad"].head()
````

Para pedir varias, pon una **lista** de nombres; por eso salen dos pares de corchetes. Con un solo par, verás un `KeyError`.

````python
ventas[["fecha", "ciudad", "total"]].head()
````

### Paso 3: filtrar filas

Imagina que en Excel agregas una columna con `=F2="Lima"`, que dice `VERDADERO` o `FALSO` en cada fila. pandas hace lo mismo con la comparación que viste en el episodio 0.3:

````python
(ventas["ciudad"] == "Lima").head()
````

`True` es verdadero y `False` es falso (por eso dice `bool`, el tipo de esos valores). Si metes esa comparación entre corchetes, pandas se queda solo con las filas `True`. Eso es un filtro:

````python
ventas_lima = ventas[ventas["ciudad"] == "Lima"]

print(ventas_lima.shape)
ventas_lima.head()
````

Los saltos en el índice son normales: cada fila conserva su número, como cuando filtras en Excel.

> ⚠️ Si tu filtro da 0 filas, no hay ningún error que te avise. Revisa mayúsculas, tildes y espacios: `"lima"` no es `"Lima"`.

Para **dos condiciones**, cada una va entre paréntesis y se unen con `&`, que equivale a `=Y()` (`AND`). No es el `&` de Excel que une textos. Para `=O()` (`OR`), se usa `|`.

> ⚠️ Aquí no sirven el `and` y el `or` del episodio 0.3: con tablas dan un `ValueError`. Y si olvidas los paréntesis, verás un `TypeError` con la palabra `rand_`.

````python
grandes_lima = ventas[(ventas["ciudad"] == "Lima") & (ventas["total"] >= 1000)]

grandes_lima.shape
````

### Paso 4: ordenar

`.sort_values()` ordena por una columna. Con `ascending=False` va de mayor a menor. Y se puede encadenar con `.head()` para ver solo las primeras filas:

````python
ventas_lima.sort_values("total", ascending=False).head()
````

Al ordenar, cada fila también conserva su número. Excel, en cambio, renumera.

> ⚠️ A diferencia de Excel, filtrar y ordenar **no cambian tu tabla original**: crean una tabla nueva. Si quieres conservarla, guárdala en una variable, como hicimos con `ventas_lima`.

### Todo junto

Las ventas más grandes de Lima, con las columnas que nos interesan. Es el pedido del inicio. Como `columnas` ya es una lista, aquí basta un par de corchetes:

````python
columnas = ["fecha", "id_producto", "cantidad", "total"]

ventas_lima.sort_values("total", ascending=False)[columnas].head()
````

## ✅ Resumen

| En Excel | En Python |
|---|---|
| Abrir un CSV | `pd.read_csv("archivo.csv")` |
| Abrir una hoja de un libro | `pd.read_excel("libro.xlsx", sheet_name="Hoja")` |
| Una hoja | Un DataFrame |
| Mirar las primeras filas | `ventas.head()` |
| Ver cuántas filas y columnas hay | `ventas.shape` |
| Quedarte con algunas columnas | `ventas[["fecha", "total"]]` |
| Filtrar | `ventas[ventas["ciudad"] == "Lima"]` |
| `=Y()` (`AND`) en un filtro | `ventas[(condición 1) & (condición 2)]` |
| `=O()` (`OR`) en un filtro | `ventas[(condición 1) \| (condición 2)]` |
| Ordenar de mayor a menor | `ventas.sort_values("total", ascending=False)` |
