### Prepara los datos

Ejecuta estas dos celdas: la primera llama a pandas y guarda la dirección de los datos, y la segunda abre las ventas.

Los ejercicios 3 y 4 usan lo que creas en el 1, y los 6 y 7, lo del 2. Si Colab se desconecta, vuelve a ejecutar todo desde arriba.

```python
import pandas as pd

DATOS = "https://raw.githubusercontent.com/PercyTechX/del-excel-al-ml/main/datos/"
```

```python
ventas = pd.read_csv(DATOS + "ventas.csv")
```

### 1. Abre los clientes

Abre el archivo `clientes.csv` en una variable llamada `clientes`. Muestra cuántas filas y columnas tiene, y sus primeras 5 filas.

> 💡 Pista: igual que en la clase, `DATOS + "clientes.csv"`. Sin `DATOS`, Python busca el archivo en la carpeta 📁 y da `FileNotFoundError`.

```python
clientes = pd.read_csv(DATOS + "clientes.csv")

print(clientes.shape)
clientes.head()
```

### 2. Abre una hoja de Excel

Abre la hoja `Productos` del libro `reporte_ventas_2025.xlsx` en una variable llamada `productos`, y muestra sus primeras 5 filas.

```python
productos = pd.read_excel(DATOS + "reporte_ventas_2025.xlsx", sheet_name="Productos")

productos.head()
```

### 3. Solo lo que necesitas

Muestra las primeras 5 filas de `clientes`, solo con las columnas `nombre`, `tipo` y `ciudad`.

```python
clientes[["nombre", "tipo", "ciudad"]].head()
```

### 4. Los clientes de Arequipa

Filtra los clientes de **Arequipa** y guárdalos en la variable `clientes_arequipa`. ¿Cuántos son?

> 💡 Pista: la primera parte de `.shape` es la cantidad de filas.

```python
clientes_arequipa = clientes[clientes["ciudad"] == "Arequipa"]

clientes_arequipa.shape
```

Son **47 clientes**, con 7 columnas.

### 5. Crédito en Cusco

Filtra las `ventas` de **Cusco** cuyo `metodo_pago` sea **Crédito**. ¿Cuántas son?

> 💡 Pista: cada condición va entre paréntesis y se unen con `&`. Si te da 0 filas, revisa la tilde de `Crédito`.

```python
credito_cusco = ventas[(ventas["ciudad"] == "Cusco") & (ventas["metodo_pago"] == "Crédito")]

credito_cusco.shape
```

Son **1,045 ventas**.

### 6. Los más caros

De la tabla `productos`, muestra los **5 con el precio más alto**, solo con las columnas `nombre` y `precio`.

```python
productos.sort_values("precio", ascending=False)[["nombre", "precio"]].head()
```

### 7. Todo junto

En la tabla `productos`, busca las **bebidas** que cuestan **menos de S/ 40**: la columna `categoria` debe ser `Bebidas` y `precio` debe ser menor que `40`. Muéstralas ordenadas de la más barata a la más cara, solo con las columnas `nombre` y `precio`.

> 💡 Pista: si no escribes `ascending=False`, `.sort_values()` ordena de menor a mayor.

```python
bebidas = productos[(productos["categoria"] == "Bebidas") & (productos["precio"] < 40)]

bebidas.sort_values("precio")[["nombre", "precio"]]
```
