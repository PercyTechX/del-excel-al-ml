## 📊 En Excel lo harías así

2. Escribes `=2+2` en una **celda** y presionas **Enter**.
3. Excel te muestra el resultado: `4`.
4. Guardas el archivo con **Ctrl + G** (`Ctrl + S` en inglés).

En Google Colab vas a hacer exactamente lo mismo. Solo cambian los nombres de algunas cosas.

## 🐍 En Python

### ¿Qué es Google Colab?

Google Colab es como un Excel que funciona en tu navegador, pero **en vez de fórmulas escribes Python**. Es es gratis y solo necesitas una cuenta de Google (la misma de Gmail).

| En Excel | En Colab |
|---|---|
| Libro (`.xlsx`) | Notebook (`.ipynb`) |
| Celda | Celda (de texto o de código) |
| Se guarda en tu computadora o en OneDrive | Se guarda en tu Google Drive |

### Paso 1: guarda tu propia copia

Este notebook está en el repositorio del curso, no en tu Drive. Si lo modificas sin guardar una copia, **perderás tus cambios** al cerrar la pestaña.

Ve al menú **Archivo → Guardar una copia en Drive**. Se abrirá una pestaña nueva con *tu* copia: trabaja siempre en esa.

> 💡 La primera vez que ejecutes una celda, Colab te mostrará una advertencia porque el notebook viene de GitHub. Haz clic en **Ejecutar de todos modos**. Luego espera unos segundos mientras Colab se conecta: arriba a la derecha verás **Conectar** y después un ✓.

### Paso 2: dos tipos de celdas

Un notebook tiene dos tipos de celdas:

- **Celdas de texto** (como esta): sirven para escribir explicaciones, títulos y notas. Son como un cuadro de texto en Excel.
- **Celdas de código**: ahí escribes instrucciones para la computadora. Son como una celda donde escribes una fórmula.

Para crear una celda nueva, usa los botones **+ Código** y **+ Texto** de la parte superior.

### Paso 3: tu primera celda de código

La celda de abajo es como escribir `=2+2` en Excel, pero **sin el signo `=`**.

Haz clic en la celda y presiona **Shift + Enter** para ejecutarla. El resultado aparece justo debajo.

```python
2 + 2
```

¡Listo, ya ejecutaste tu primer código en Python! 🎉

Las operaciones son las mismas que usas en Excel:

| Operación | En Excel | En Python |
|---|---|---|
| Sumar | `=10+5` | `10 + 5` |
| Restar | `=10-5` | `10 - 5` |
| Multiplicar | `=10*5` | `10 * 5` |
| Dividir | `=10/5` | `10 / 5` |
| Paréntesis | `=(10+5)/2` | `(10 + 5) / 2` |

Por ejemplo, si ganas **S/ 2,500** al mes, ¿cuánto ganas en un año?

```python
2500 * 12
```

Y el promedio de ventas de tres días:

```python
(1250.50 + 980 + 1430) / 3
```

Python muestra muchos decimales cuando el resultado no es exacto (o un `.0` cuando hay decimales de por medio). En el episodio 0.2 veremos cómo redondear, como con `=REDONDEAR()` (`ROUND`).

### ⚠️ Cuidado con las comas

En Python los decimales se escriben con **punto** (`1250.50`) y los miles van **sin separador**. Mira qué pasa si escribimos el sueldo como en Excel, con coma:

```python
2,500 * 12
```

Python vio **dos números separados** (`2` y `500`) y solo multiplicó el segundo. Por eso los miles se escriben **sin coma**: `2500`.

### Mostrar un mensaje

Para mostrar un texto usamos `print()`. El texto va entre comillas, igual que cuando escribes `="Hola"` en una celda de Excel.

Lo que va después de `#` es un **comentario**: una nota para las personas, que Python ignora.

```python
# Esta línea es un comentario: Python no la ejecuta
print("¡Hola, Distribuidora Andina!")
```

### El orden importa

En Excel, si cambias una celda, todo se recalcula solo. **En Colab no**: cada celda se ejecuta cuando tú lo decides, en el orden que tú elijas.

El número entre corchetes a la izquierda de cada celda, como `[3]`, indica en qué orden se ejecutó.

Mira este ejemplo. En la primera celda le ponemos un nombre a un valor, como cuando le pones nombre a una celda en Excel (lo veremos a fondo en el episodio 0.2):

```python
precio = 25
```

```python
precio * 4
```

Si cambias `precio = 25` por `precio = 30`, **tienes que volver a ejecutar las dos celdas** para ver el nuevo resultado.

> 💡 Si algo se ve raro, usa **Entorno de ejecución → Reiniciar sesión y ejecutar todo**. Es como empezar el libro desde cero.

### Paso 4: los archivos

A la izquierda de Colab hay un ícono de carpeta 📁. Ahí están los archivos con los que trabaja tu notebook.

**Para subir un archivo tuyo** (por ejemplo, un Excel de tu trabajo):

1. Abre la carpeta 📁 de la izquierda.
2. Haz clic en el ícono de **subir** (una hoja con una flecha hacia arriba) y elige el archivo de tu computadora. También puedes **arrastrarlo** a la carpeta.
3. El archivo aparecerá en la lista.

> ⚠️ **Importante:** los archivos que subes a esa carpeta son **temporales**: se borran cuando Colab cierra tu sesión, por ejemplo si lo dejas sin usar un buen rato. El notebook sí queda guardado en tu Drive, pero esos archivos no. En los próximos episodios veremos cómo trabajar directamente con archivos de tu Google Drive.
