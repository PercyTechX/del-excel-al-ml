### 1. Las ventas de la semana

```python
1250 + 980 + 1430
```

### 2. El promedio diario

Calcula el **promedio** de ventas de esos tres días.

> 💡 Pista: igual que en Excel, usa paréntesis para sumar primero y luego dividir.

```python
(1250 + 980 + 1430) / 3
```

### 3. El IGV

Una factura tiene un valor de venta de **S/ 1,500**. Calcula el **IGV (18 %)**.

> 💡 Pista: el 18 % se escribe `0.18`.

```python
1500 * 0.18
```

### 4. El precio con IGV

Calcula el **importe total** de esa factura: el valor de venta más el IGV.

```python
1500 + 1500 * 0.18
```

Otra forma más corta: multiplicar por `1.18`, porque 100 % + 18 % = 118 %.

```python
1500 * 1.18
```
